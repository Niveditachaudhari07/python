## Topic 3: Nested Dictionaries

1. Given `user = {"profile": {"name": "Mina"}}`, print the name.
2. Add `email` inside the nested profile.
3. Use `.get()` to read a missing `phone` field with default `"not provided"`.

```python
user = {
    "profile": {
        "name": "Mina"
    }
}

print(user["profile"]["name"])
```

```python
user = {
    "profile": {
        "name": "Mina"
    }
}

user["profile"]["email"] = "mina@gmai.com"

print(user)
```

```python
user = {
    "profile": {
        "name": "Mina",
        "email": "mina@gmail.com"
    }
}

phone = user["profile"].get("phone", "not provided")

print(phone)
```

## Topic 4: Conditional Statements

1. Classify `score = 72` as `pass` if it is at least 50, otherwise `fail`.
2. Apply a 10% discount when `amount > 200`, otherwise apply 5%.
3. Print `eligible` only when `age >= 18 and has_id is True`.

```python
score = 72

if score >= 50:
    print("pass")
else:
    print("fail")
```

```python
amount = 250

if amount > 200:
    discount = amount * 0.10
else:
    discount = amount * 0.05

final_amount = amount - discount

print(final_amount)
```

```python
age = 25
has_id = True

if age >= 18 and has_id is True:
    print("eligible")
```

## Topic 5: Loops

1. Calculate the total of `[25, 40, 15, 20]`.
2. Count values at least 20 in the same list.
3. Print each item in `['paid', 'pending', 'paid']` with its position using `enumerate()`.

```python
numbers = [25, 40, 15, 20]

total = 0

for number in numbers:
    total = total + number

print(total)
```

```python
numbers = [25, 40, 15, 20]

count = 0

for number in numbers:
    if number >= 20:
        count = count + 1

print(count)
```

```python
statuses = ["paid", "pending", "paid"]

for position, status in enumerate(statuses):
    print(position, status)
```

## Topic 6: Functions

1. Write `square(number)` that returns the number squared.
2. Write `full_name(first, last)` that returns a combined string.
3. Write `shipping_cost(amount, free_limit=100)` that returns 0 when the amount reaches the limit, otherwise 10.

```python
def square(number):
    return number * number

print(square(5))
```

```python
def full_name(first, last):
    return first + " " + last

print(full_name("nive", "bella"))
```
