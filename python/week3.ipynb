## Topic 5: Loops

1. Calculate total revenue from a list of order dictionaries.
2. Count paid orders and collect the IDs of unpaid orders.
3. Validate records in a loop and collect IDs whose amount is missing or non-positive.

```python
orders = [
    {'id': 1, 'amount': 100, 'status': 'paid'},
    {'id': 2, 'amount': 50, 'status': 'unpaid'},
    {'id': 3, 'amount': 200, 'status': 'paid'},
    {'id': 4, 'amount': 0, 'status': 'unpaid'}
]

total_revenue = 0

for order in orders:
    if order['status'] == 'paid':
        total_revenue += order['amount']

print(total_revenue)
```

```python
paid_count = 0
unpaid_ids = []

for order in orders:
    if order['status'] == 'paid':
        paid_count += 1
    else:
        unpaid_ids.append(order['id'])

print(paid_count)
print(unpaid_ids)
```

```python
records = [
    {'id': 1, 'amount': 100},
    {'id': 2, 'amount': 0},
    {'id': 3},
    {'id': 4, 'amount': -20}
]

invalid_ids = []

for record in records:
    amount = record.get('amount')

    if amount is None or amount <= 0:
        invalid_ids.append(record['id'])

print(invalid_ids)
```

## Topic 6: Functions

1. Write `net_price(price, discount_rate=0)` with validation for a rate from 0 to 1.
2. Write `active_employee_names(records)` that returns names for active employees.
3. Write `summarize_values(values)` returning count, total, and average.

```python
def net_price(price, discount_rate=0):
    if price < 0:
        raise ValueError('Price cannot be negative')

    if discount_rate < 0 or discount_rate > 1:
        raise ValueError('Discount rate must be between 0 and 1')

    return price * (1 - discount_rate)


print(net_price(100, 0.2))
print(net_price(100))
```

```python
def summarize_values(values):
    count = len(values)
    total = sum(values)

    if count > 0:
        average = total / count
    else:
        average = 0

    return {
        'count': count,
        'total': total,
        'average': average
    }
print(summarize_values([10, 20, 30]))
```
