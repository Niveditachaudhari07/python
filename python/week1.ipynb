## Topic 4: Conditional Statements

### Question 1
If `temperature = 25`, print `warm` when it is above 20; otherwise print `cool`.

```python
temperature = 25

if temperature > 20:
    print("warm")
else:
    print("cool")
```

## Topic 5: Loops

### Question 1
Use a `for` loop to print each item in `[10, 20, 30]`.

```python
numbers = [10, 20, 30]

for number in numbers:
    print(number)
```

### Question 2
Use a loop to calculate the sum of `[2, 4, 6]`.

```python
numbers = [2, 4, 6]

total = 0

for number in numbers:
    total = total + number

print(total)
```

### Question 3
Count how many values in `[5, 12, 8, 20]` are greater than 10.

```python
numbers = [5, 12, 8, 20]

count = 0

for number in numbers:
    if number > 10:
        count = count + 1

print(count)
```
