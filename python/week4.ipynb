## Topic 1: Data Types

1. Normalize a record whose quantity and price arrive as strings; reject missing or invalid values.
2. Distinguish `None`, `0`, `False`, and an empty string in a small reporting function.
3. Parse a mixed list into separate integer, float, and invalid-value lists without raising an exception.

```python
order = {
    'quantity': '3',
    'price': '14.50'
}

def normalize_order(order):
    try:
        quantity = int(order['quantity'])
        price = float(order['price'])

        if quantity <= 0 or price < 0:
            return None

        return {
            'quantity': quantity,
            'price': price,
            'total': quantity * price
        }

    except (KeyError, TypeError, ValueError):
        return None


order = {'quantity': '3', 'price': '14.50'}

result = normalize_order(order)

print(result)
```

```python
def report_value(value):
    if value is None:
        return 'Missing value'

    elif value is False:
        return 'False value'

    elif value == 0:
        return 'Zero'

    elif value == '':
        return 'Empty string'

    else:
        return 'Value supplied'


print(report_value(None))
print(report_value(0))
print(report_value(False))
print(report_value(''))
print(report_value(100))
```

## Topic 4: Conditional Statements

1. Implement tiered discounts: 0% below 100, 5% from 100, and 10% from 500.
2. Route a record to `accepted`, `review`, or `rejected` based on amount, status, and required fields.
3. Use compound conditions to identify high-value paid orders from a list of records.

```python
amount = 600

if amount < 100:
    discount_rate = 0
elif amount < 500:
    discount_rate = 0.05
else:
    discount_rate = 0.10

discount = amount * discount_rate
net_amount = amount - discount

print('Discount:', discount)
print('Net amount:', net_amount)
```

```python
def route_order(order):
    if 'id' not in order or 'amount' not in order or 'status' not in order:
        return 'rejected'

    try:
        amount = float(order['amount'])
    except (ValueError, TypeError):
        return 'rejected'

    if amount <= 0 or order['status'] != 'paid':
        return 'rejected'
    elif amount > 1000:
        return 'review'
    else:
        return 'accepted'


print(route_order({'id': 1, 'amount': 100, 'status': 'paid'}))
print(route_order({'id': 2, 'amount': 1500, 'status': 'paid'}))
print(route_order({'id': 3, 'amount': 0, 'status': 'paid'}))
```

## Topic 5: Loops

1. Compute revenue by category from order dictionaries.
2. Process records while collecting valid outputs and an error list with record IDs.
3. Use nested loops to calculate each student's average score and identify the highest average.

```python
orders = [
    {'category': 'Electronics', 'amount': 500},
    {'category': 'Clothing', 'amount': 100},
    {'category': 'Electronics', 'amount': 300},
    {'category': 'Clothing', 'amount': 200}
]

revenue = {}

for order in orders:
    category = order['category']
    amount = order['amount']

    if category not in revenue:
        revenue[category] = 0

    revenue[category] += amount

print(revenue)
```

```python
records = [
    {'id': 1, 'amount': 100},
    {'id': 2, 'amount': 0},
    {'id': 3},
    {'id': 4, 'amount': 250}
]

valid_records = []
errors = []

for record in records:
    amount = record.get('amount')

    if amount is None:
        errors.append({'id': record.get('id'), 'error': 'Missing amount'})
    elif amount <= 0:
        errors.append({'id': record.get('id'), 'error': 'Amount must be positive'})
    else:
        valid_records.append(record)

print('Valid:', valid_records)
print('Errors:', errors)
```

```python
students = {
    'Asha': [80, 90, 70],
    'Raj': [60, 70, 80],
    'Mina': [90, 95, 85]
}

averages = {}

for student, scores in students.items():
    total = 0

    for score in scores:
        total += score

    average = total / len(scores)
    averages[student] = average

print('Averages:', averages)

best_student = max(averages, key=averages.get)

print('Highest average:', best_student, averages[best_student])
```

## Topic 6: Functions

1. Create a pure `validate_order(order)` function returning `(is_valid, errors)`.
2. Create `transform_order(order)` that returns a new record with a calculated net amount.
3. Compose validation and transformation into `process_orders(orders)` returning successes and failures.

```python
def validate_order(order):
    errors = []

    if 'id' not in order:
        errors.append('Missing order ID')

    if 'amount' not in order:
        errors.append('Missing amount')
    else:
        try:
            amount = float(order['amount'])

            if amount <= 0:
                errors.append('Amount must be positive')

        except (ValueError, TypeError):
            errors.append('Amount must be numeric')

    is_valid = len(errors) == 0

    return is_valid, errors


print(validate_order({'id': 1, 'amount': 100}))
print(validate_order({'id': 2, 'amount': 0}))
```

```python
def transform_order(order):
    new_order = order.copy()

    amount = float(order['amount'])
    discount_rate = 0.10

    new_order['net_amount'] = amount * (1 - discount_rate)

    return new_order


order = {'id': 1, 'amount': 100}

result = transform_order(order)

print(result)
print(order)
```

```python
def process_orders(orders):
    successes = []
    failures = []

    for order in orders:
        is_valid, errors = validate_order(order)

        if is_valid:
            transformed = transform_order(order)
            successes.append(transformed)
        else:
            failures.append({
                'order': order,
                'errors': errors
            })

    return successes, failures


orders = [
    {'id': 1, 'amount': 100},
    {'id': 2, 'amount': 0},
    {'id': 3, 'amount': 250}
]

successes, failures = process_orders(orders)

print('Successes:', successes)
print('Failures:', failures)
```
